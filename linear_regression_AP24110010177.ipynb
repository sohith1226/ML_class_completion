This notebook fits a straight-line model to predict house price from area.


```python
# Import libraries for tabular data, plotting, and linear regression.
# This cell performs the next step of the analysis.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Create a small example dataset; price is measured in lakhs.
#CREATING THE DATASET FOR THE CODE - USING THE HOUSE RENT AS AN EXAMPLE FOR IT
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Price': [30, 38, 45, 48, 80, 60, 80, 95]
}

df = pd.DataFrame(data)
print(df)

X = df[['Area']]
y = df['Price']
```

```python
# Fit the linear regression model using area as the input feature.
#TO TRAIN THE MODEL WE USE
model = LinearRegression()
model.fit(X, y)

print("Model trained successfully")
```

```python
# The coefficient is the slope; the intercept is the predicted value at area = 0.
# This cell performs the next step of the analysis.
print("Slope:", round(model.coef_[0], 2))
print("Intercept:", round(model.intercept_, 2))
```

```python
# Compare observed prices with prices predicted for the training examples.
# TO PREDICT THE PRICES
y_pred = model.predict(X)

print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
# Use the trained model to estimate the price of a house with area 2.4.
#PREDICT A NEW HOUSE PRICES

new_area = pd.DataFrame({'Area': [2.4]})
price = model.predict(new_area)
print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
# Plot the observed prices and the fitted regression line.
# to get the regression line

plt.scatter(X, y, color='blue', label='Actual Prices')
plt.plot(X, y_pred, color='red', label='Regression Line')

plt.xlabel("Area")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```
