##COVARIANCE:

**IT IS A MEASURE OF HOW 2 VARIABLES CHANGING THEIR DIRECTION OF THEIR LINEAR RELATIONSHIP**

=> IT CONTAINS HUGE DATA AND IT CAN BE MEASURED USING COVARIANCE MATRIX WHERE THEY ARE EITHER RELATED OR NON-RELATED.

* +VE COVARIANCE IS - IF 1 VARIABLE INCREASE THE OTHER ONE INCREASES TOO
* 0 COVARIANCE - HAPPEN WHEN 2 VARIABLE SWERE NOT RELATE DSO NO EFFECT.


**FORMULA** = 1 / n - 1 * E OF K = 1 TO  n OF (X - x)(Y - y)



```python

X1 = [4, 8  ,13, 7]
Y1 = [11, 4, 5, 14]
mean_X = sum(X1) / len(X1)
mean_Y = sum(Y1) / len(Y1)
diff_X = [x - mean_X for x in X1]
diff_Y = [y - mean_Y for y in Y1]

products = [diff_X[i] * diff_Y[i] for i in range(len(X1))]
sum_products = sum(products)

covariance = sum_products / len(X1)
print(f"Covariance between X1 and Y1: {covariance}")
```

```python
X = [4, 8, 13, 7]
Y = [11, 4, 5, 14]

mean_X = sum(X) / len(X)
mean_Y = sum(Y) / len(Y)

diff_X = [x - mean_X for x in X]
diff_Y = [y - mean_Y for y in Y]

products = [diff_X[i] * diff_Y[i] for i in range(len(X))]
sum_products = sum(products)

covariance = sum_products / len(X)

print(f"Mean of X: {mean_X}")
print(f"Mean of Y: {mean_Y}")
print(f"Differences from Mean (X): {diff_X}")
print(f"Differences from Mean (Y): {diff_Y}")
print(f"Products of differences: {products}")
print(f"Sum of products: {sum_products}")
print(f"Covariance between X and Y: {covariance}")
```

### TO GET THE COVARIANCE MATRIX:

 WE USE COV = [ COV(X1,X1)    COV(X1,Y1)  COV(Y1, X1)  COV(Y1,Y1) ]

```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
X1 = [4, 8, 13, 7]
Y1 = [11, 4, 5, 14]
data = np.array([X1, Y1])

# Calculate covariance matrix
cov_matrix = np.cov(data)

#  covariance matrix
print("Covariance Matrix:\n", cov_matrix)

# Plotting the covariance matrix as a heatmap
plt.figure(figsize=(5, 4))
sns.heatmap(
    cov_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    xticklabels=["X1", "Y1"],
    yticklabels=["X1", "Y1"]
)

plt.title("Covariance Matrix Heatmap")
plt.show()
```

## CORRELATION:


IN COVARIANCE WE CAN FIND THE RELATIONSHIP B/W 2 VARIABLE BUT IT CANT GIVE THE STRENGTH OF REALTIO SO CORRELATIONS COMES IN.IT HELPS TO GIVE THE STRENGTH OF THE RELATION .

* THE VALUES ARE B/W +1 TO -1 AND WHEN IT IS NEAR T0 EITHER OF THE ONE IT IS CLASSIFIED AS EITHER POSITIVE OR NEGATIVE LINEAR RELATION AND ID IT IS ZERO IT WILL BE DECIDED AS NO RELATION.

* CORRELATION = COV ( X,Y ) / SIGMA (x) * SIGMA (y)

- FOR COR = 1 IT IS PERFECT REALTION.
-  FOR COR = 0 NO REALTION
- FOR COR = > 0.5 NEAR TO GOOD RELATION
- FOR COR = < 0.5 WEAK RELATION   

```python
X = [2, 4, 6, 8, 10]
Y = [50, 60, 70, 80, 90]

mean_X = sum(X) / len(X)
mean_Y = sum(Y) / len(Y)

numerator = sum((X[i] - mean_X) * (Y[i] - mean_Y) for i in range(len(X)))

sum_X = sum((x - mean_X) ** 2 for x in X)
sum_Y = sum((y - mean_Y) ** 2 for y in Y)

correlation = numerator / (sum_X * sum_Y) ** 0.5

print("Mean of X:", mean_X)
print("Mean of Y:", mean_Y)
print("Correlation:", correlation)
```

### **TAKING A DATASET OF HOSUE PRICING AND FINDING THE CORELATION AND COVARIANCE FOR IT ⁉**


```python
area = [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8]
rooms = [2, 2, 3, 3, 4, 4, 5, 5]
age = [10, 15, 18, 12, 14, 18, 16, 15]
distance = [8, 4, 3, 5, 1, 4, 6, 3]
price = [30, 38, 45, 48, 80, 60, 80, 95]

# TO calculate covariance and correlation
def calculate(x, y):
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)
    # FINIDNG THE  DIFFERENCE FROM THE MEAN
    dx = [i - mean_x for i in x]
    dy = [i - mean_y for i in y]
    #  covariance
    covariance = sum(dx[i] * dy[i] for i in range(len(x))) / len(x)
    # squared differences
    sx = sum(i ** 2 for i in dx)
    sy = sum(i ** 2 for i in dy)
    #  correlation
    correlation = sum(dx[i] * dy[i] for i in range(len(x))) / (sx * sy) ** 0.5
    return covariance, correlation

#  results for each variable against price
for name, values in [("Area", area), ("Rooms", rooms),
                     ("Age", age), ("Distance", distance)]:
    cov, corr = calculate(values, price)
    print(f"{name} - Covariance: {cov:.2f}, Correlation: {corr:.4f}")
```

# Principal Component Analysis (PCA):

Uses dimensionality reduction technique and helps us to reduce the number of features in a dataset while keeping the most important information.

- give the correct output even after having similar data i more than 1/2 rows of the dataset

**how to calculate**:
- **step - 1 : SD (TO COMPARE CORRRECTLY THE PCA FIRST STANDERD VERIOSN VERIFICATION WILL BE DONW.)'
- FORMULAR =  X - MEAN OF INDEPENDENT VAR. / SD**

- ***STEP  - 02: CALCULATE THE COVARIANCE ( TO CHCECK how features relate to each other)**

- **STEP - 03: Find the Principal Components**
* The direction of maximum variance - FOR PC1
* The next best direction, perpendicular to PC1 and so on- FOR PC2

- **STEP - 04: Top Directions & Transform Data**


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

data = {
    'House': [1, 2, 3, 4, 5, 6, 7, 8],
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'No. of Rooms': [2, 2, 3, 3, 4, 4, 5, 5],
    'Age': [10, 15, 18, 12, 14, 18, 16, 15],
    'Distance': [8, 4, 3, 5, 1, 4, 6, 3],
    'Price': [30, 38, 45, 48, 80, 60, 80, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
X = df.drop('Price', axis=1)
y = df['Price']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled Features (X):")
print(X_scaled)

print("\\nTarget Variable (y):")
print(y)
```

```python
# Apply PCA with 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Display PCA results
pca_df = pd.DataFrame(
    X_pca,
    columns=['Principal Component 1', 'Principal Component 2']
)

print("PCA Transformed Data:")
print(pca_df)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Explained:")
print(pca.explained_variance_ratio_.sum())
```
