```python

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}
#AP24110010760
df = pd.DataFrame(data)

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

print("Original Data:")
print(df)

print("\nPCA Transformed Data:")
print(pca_data)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Preserved:")
print(sum(pca.explained_variance_ratio_))

```

Principal Component Analysis (PCA)

program implements Principal Component Analysis (PCA), a dimensionality reduction technique used in machine learning. First, it imports Pandas, StandardScaler, and PCA from Scikit-learn. A dataset containing multiple numerical attributes is created and converted into a DataFrame. The StandardScaler() function standardizes the data so that the features have comparable scales. Next, PCA(n_components=2) selects two principal components, and fit_transform() transforms the standardized data into a reduced-dimensional representation. The program then displays the original and transformed datasets. Finally, explained_variance_ratio_ calculates the proportion of variance explained by each principal component, and the sum of these values indicates the total variance preserved by the selected components.
