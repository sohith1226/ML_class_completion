## Multiple Linear Regression — AP24110010177

This notebook predicts house price from multiple house-related features.


```python
# Import tools for data handling, visualization, and regression.
# This cell performs the next step of the analysis.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Example dataset with several possible factors affecting house price.
# This cell performs the next step of the analysis.
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Rooms': [2, 2, 3, 3, 4, 4, 5, 5],
    'Age': [10, 15, 18, 12, 14, 18, 16, 15],
    'Distance': [8, 4, 3, 5, 1, 4, 6, 3],
    'Price': [30, 38, 45, 48, 80, 60, 80, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Separate input features (X) from the target price (y).
#TO GET THE I/P AND O/P:
X = df[['Area', 'Rooms', 'Age', 'Distance']]
y = df['Price']
```

```python
# Train the model to estimate price using all input features together.
# This cell performs the next step of the analysis.
model = LinearRegression()
model.fit(X, y)
print("Model trained successfully")
```

```python
# Each coefficient describes the model contribution of a feature; intercept is the baseline.
# This cell performs the next step of the analysis.
print("Coefficients:", np.round(model.coef_, 2))
print("Intercept:", round(model.intercept_, 2))
```

```python
# Predict prices for the houses already present in the example dataset.
#PREDICTING THE HOUSE PRICES

y_pred = model.predict(X)

print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
# Predict the price for one new house using its four feature values.
#PREDICTING A NEW HOUSE PRICE
new_house = pd.DataFrame({
    'Area': [2.4],
    'Rooms': [4],
    'Age': [12],
    'Distance': [3]
})

price = model.predict(new_house)

print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
# R-squared summarizes how well the model fits this same dataset.
# TO CALCULATE R^2:

print("R2 Score:", round(model.score(X, y), 2))
```
