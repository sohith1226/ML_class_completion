## Polynomial Regression — AP24110010177

This notebook models a potentially curved relationship between house area and price using degree-2 features.


```python
# Import data tools, a linear estimator, and the polynomial feature transformer.
# This cell performs the next step of the analysis.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
# Create the example dataset of house area and price.
# This cell performs the next step of the analysis.
data = {
    'Area': [1.0, 1.2, 1.5, 1.8, 2.0, 2.2, 2.5, 2.8],
    'Price': [30, 38, 45, 48, 80, 60, 80, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Set area as the input and price as the value to predict.
# This cell performs the next step of the analysis.
X = df[['Area']]
y = df['Price']
```

```python
# Expand area into polynomial terms, including area squared, for a curved fit.
#CONVERTING THE DATA INTO A PEOLYNOMIAL FEATURE

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

print(X_poly)
```

```python
# Fit linear regression to the transformed polynomial features.
# This cell performs the next step of the analysis.
model = LinearRegression()
model.fit(X_poly, y)

print("Model trained successfully")
```

```python
# Display learned polynomial coefficients and the intercept.
#DISPLAYING OUTPUTVALUES
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
# Compare actual prices with predictions from the fitted model.
#FOR THE SAKE OF PREDICTING THE HOUSE'S PRICES:

y_pred = model.predict(X_poly)

print("Actual Prices:", y.values)
print("Predicted Prices:", y_pred.round(2))
```

```python
# Transform the new area with the same transformer before predicting its price.
#PREDITING FOR THE NEW HOUSE

new_area = pd.DataFrame({'Area': [2.4]})

new_area_poly = poly.transform(new_area)
price = model.predict(new_area_poly)

print("Predicted House Price:", round(price[0], 2), "Lakhs")
```

```python
# Plot the observed data and the model predictions.
#TO PLOT THE CURVE
plt.scatter(X, y, color='blue', label='Actual Prices')

plt.plot(X, y_pred, color='red', label='Polynomial Curve')

plt.xlabel("Area")
plt.ylabel("Price (Lakhs)")
plt.title("Polynomial Regression")
plt.legend()
plt.show()
```

```python
# Calculate R-squared on the examples used to fit the model.
#TO GET THE R^2 VALUE WE USE :
print("R2 Score:", round(model.score(X_poly, y), 2))
```
